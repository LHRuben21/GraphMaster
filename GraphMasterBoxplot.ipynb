```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Simular datos de un experimento (n=30 por grupo)
np.random.seed(42)   # semilla: hace que datos aleatorios sean reproducibles

datos = pd.DataFrame({
    'Grupo': ['Control']*30 + ['Tto A']*30 + ['Tto B']*30,
    'Valor': np.concatenate([
        np.random.normal(45, 10, 30),   # media=45, sd=10
        np.random.normal(68, 8, 30),
        np.random.normal(82, 6, 30)
    ])
})

# Crear el boxplot con Seaborn
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(
    data=datos,
    x='Grupo',
    y='Valor',
    palette='colorblind',   # accesible para daltónicos
    width=0.5,
    flierprops=dict(marker='o', markersize=5)   # estilo de outliers
)

ax.set_ylabel('Respuesta celular (%)', fontsize=12)
ax.set_title('Figura 2. Distribución de respuesta por grupo')
plt.tight_layout()
plt.savefig('figura2_boxplot.pdf', dpi=300)
```
