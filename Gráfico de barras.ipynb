```
import matplotlib.pyplot as plt
import numpy as np

# 1. Definir los datos del estudio
grupos = ['Control', 'Tratamiento A', 'Tratamiento B']
medias = [100, 67.8, 82.4]
errores = [8.3, 6.1, 5.2]   # Desviación estándar

# 2. Crear la figura y los ejes
fig, ax = plt.subplots(figsize=(7, 5))

# 3. Dibujar las barras
x = np.arange(len(grupos))  # posiciones en eje X: [0, 1, 2]
bars = ax.bar(x, medias,
              yerr=errores,
              capsize=5,
              color=['#4878D0', '#EE854A', '#6ACC65'],
              edgecolor='black',
              linewidth=0.8)

# 4. Etiquetas y título (OBLIGATORIOS para journals)
ax.set_ylabel('Respuesta celular (%)', fontsize=12)
ax.set_xlabel('Grupo experimental', fontsize=12)
ax.set_title('Figura 1. Respuesta celular por grupo', fontsize=13)
ax.set_xticks(x)
ax.set_xticklabels(grupos)
ax.set_ylim(0, 100)

# 5. Guardar en alta resolución para publicar
plt.tight_layout()
plt.savefig('figura1.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
